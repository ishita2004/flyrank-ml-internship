## 2. The question: decision, action, cost of a wrong call

- **The Decision:** Which existing content pages should be prioritized first for a content refresh review?
- **Who Acts & The Action:** SEO managers and Content Strategy teams. Based on the ranked queue and reason codes (e.g. `stale_visible_page`, `declining_with_demand`), editors inspect the flagged pages to update outdated statistics, refresh internal links, expand thin sections, or improve title/meta CTR.
- **Cost of a Wrong Call:**
  - *False Positive (recommending a low-upside page):* Wastes expensive editorial hours and content budget on pages with minimal search volume or structural lack of demand.
  - *False Negative (missing a decaying high-demand page):* Allows high-converting or high-traffic organic assets to silently lose search position and click share to competitors.
- **Why ML/Data Helps:** Simple single-threshold rules (e.g. `days_since_last_update > 180`) yield massive lists with low precision. A machine learning model combines non-linear interactions across visibility (`impressions_90d`), engagement (`ctr`, `engagement_rate`), position decay (`avg_position`), and freshness to rank candidates efficiently by expected return on review time.

## 3. Quick look at the data (2-3 real numbers)

To confirm that this lane offers real leverage on the starter dataset (`data/raw/content_refresh_anonymized.csv`), we compute three key numbers in the code cell below:
1. **Stale Visible Pages:** Pages with high search demand (`impressions_90d >= 500`) that have not been updated in 6+ months (`days_since_last_update >= 180`).
2. **Decline Rate in Stale Pages:** The proportion of these stale, visible pages whose search trend is actively declining (`trend_direction == "down"`).
3. **High-Impression Opportunity Share:** Total impressions captured by declining visible pages compared to the overall dataset.

```python
import os, sys, pandas as pd, numpy as np

# Ensure kernel is at repo root
while not os.path.isdir('data/raw') and os.getcwd() != os.path.abspath(os.sep):
    os.chdir('..')

# Load starter dataset
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# 1. Visible & Stale Pages
visible = df[df['impressions_90d'] >= 500]
stale_visible = visible[visible['days_since_last_update'] >= 180]
stale_pct = len(stale_visible) / len(visible) * 100

# 2. Active Decline Rate among Stale Visible Pages
declining_stale = stale_visible[stale_visible['trend_direction'].str.lower() == 'down']
decline_rate = len(declining_stale) / len(stale_visible) * 100

# 3. Total Impressions at Risk
declining_impressions = declining_stale['impressions_90d'].sum()
total_impressions = df['impressions_90d'].sum()
at_risk_share = (declining_impressions / total_impressions) * 100

print(f'Total Dataset Rows: {len(df):,}')
print(f'1. High-Visibility Pages (impressions_90d >= 500): {len(visible):,}')
print(f'   -> Stale Pages (last update >= 180 days): {len(stale_visible):,} ({stale_pct:.1f}% of visible pages)')
print(f'2. Active Decline Rate in Stale Visible Pages: {decline_rate:.1f}% ({len(declining_stale):,} pages)')
print(f'3. Impressions at Risk from Declining Stale Content: {declining_impressions:,} ({at_risk_share:.1f}% of total search impressions)')
```
